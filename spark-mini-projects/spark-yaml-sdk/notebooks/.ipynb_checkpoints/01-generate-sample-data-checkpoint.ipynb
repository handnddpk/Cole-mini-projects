# Generate Sample Orders Data

This notebook generates sample orders data and uploads it to MinIO (S3-compatible storage).

```python
import os
import random
from datetime import datetime, timedelta
import pandas as pd
from pyspark.sql import SparkSession
```

```python
# Initialize Spark session
spark = SparkSession.builder \
    .appName("GenerateSampleData") \
    .config("spark.hadoop.fs.s3a.endpoint", "http://minio:9000") \
    .config("spark.hadoop.fs.s3a.access.key", "minioadmin") \
    .config("spark.hadoop.fs.s3a.secret.key", "minioadmin") \
    .config("spark.hadoop.fs.s3a.path.style.access", "true") \
    .config("spark.hadoop.fs.s3a.impl", "org.apache.hadoop.fs.s3a.S3AFileSystem") \
    .config("spark.hadoop.fs.s3a.aws.credentials.provider", "org.apache.hadoop.fs.s3a.SimpleAWSCredentialsProvider") \
    .getOrCreate()
```

## Create MinIO bucket

First, let's ensure our bucket exists in MinIO. We'll use the AWS SDK for this.

```python
import boto3
from botocore.client import Config

# Create a connection to MinIO
s3_client = boto3.client(
    's3',
    endpoint_url='http://minio:9000',
    aws_access_key_id='minioadmin',
    aws_secret_access_key='minioadmin',
    config=Config(signature_version='s3v4'),
    region_name='us-east-1'
)

# Create the bucket if it doesn't exist
try:
    s3_client.head_bucket(Bucket='data-warehouse')
    print("Bucket 'data-warehouse' already exists")
except:
    s3_client.create_bucket(Bucket='data-warehouse')
    print("Created bucket 'data-warehouse'")
```

## Generate Sample Orders Data

Now let's generate some sample orders data that matches our schema.

```python
# Generate sample orders data
def generate_orders(num_orders=100):
    orders = []
    start_date = datetime(2022, 1, 1)
    end_date = datetime(2023, 1, 1)
    date_range = (end_date - start_date).days
    
    for i in range(1, num_orders + 1):
        customer_id = random.randint(1, 10)  # Match with our 10 sample customers
        product_id = random.randint(1, 10)   # Match with our 10 sample products
        order_date = start_date + timedelta(days=random.randint(0, date_range))
        quantity = random.randint(1, 5)
        price = round(random.uniform(10, 1500), 2)  # Random price between $10 and $1500
        
        orders.append({
            "order_id": i,
            "customer_id": customer_id,
            "product_id": product_id,
            "order_date": order_date.strftime("%Y-%m-%d"),
            "quantity": quantity,
            "price": price
        })
    
    return pd.DataFrame(orders)
```

```python
# Generate 200 sample orders
orders_df = generate_orders(200)
orders_df.head()
```

## Write Data to MinIO

Now let's write the data to MinIO in Parquet format.

```python
# Convert to Spark DataFrame
pd.DataFrame.iteritems = pd.DataFrame.items
spark_orders = spark.createDataFrame(orders_df)

# Show the schema
spark_orders.printSchema()
```

```python
# Write to MinIO as Parquet
spark_orders.write \
    .format("parquet") \
    .mode("overwrite") \
    .save("s3a://data-warehouse/orders/")
```

## Verify the Data

Let's read back the data to verify it was written correctly.

```python
# Read the data back from MinIO
orders_from_minio = spark.read.parquet("s3a://data-warehouse/orders/")

# Show a few records
orders_from_minio.show(5)
```

```python
# Count the records
print(f"Total orders: {orders_from_minio.count()}")
```

## Data Summary

Let's look at some summary statistics for our generated data.

```python
# Summary statistics
orders_from_minio.describe().show()
```

```python
# Distribution of orders by customer
orders_from_minio.groupBy("customer_id").count().orderBy("customer_id").show()
```

```python
# Distribution of orders by product
orders_from_minio.groupBy("product_id").count().orderBy("product_id").show()
```
